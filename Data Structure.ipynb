# A data structure is a way of organizing, storing, and managing data so that it can be accessed and modified efficiently.

## Python provides several built-in data structures.

1. List []- A list is an ordered and mutable collection that allows duplicate values.
2. Tuple () - A tuple is an ordered but immutable collection which allows duplicate values.
3. Set {}- A set is an unordered collection of unique elements.
4. Dictionary {Key:Value} - A dictionary stores data as key-value pairs.
