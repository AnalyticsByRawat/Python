```python
## Check whether a number is even or odd using the % operator.

number = int(input("Enter Number to be checked :"))

if (number%2==0):
    print("Number is Even")
else:
    print("Number is Odd")
```

```python
# Find the largest of three numbers using comparison operators.

num1 = int(input("Enter 1st number :"))

num2 = int(input("Enter 2nd number :"))

num3 = int(input("Enter 3rd number :"))

if (num1 > num2) and (num1 > num3):
    print("1st number is greater")
elif (num2 > num1) and (num2 > num3):
    print("2nd number is greater")
else:
    print("3rd number is greater")
```

```python
## Calculate a student's percentage and determine pass/fail using logical operators.

# Input marks of 5 subjects
sub1 = float(input("Enter marks for Subject 1: "))
sub2 = float(input("Enter marks for Subject 2: "))
sub3 = float(input("Enter marks for Subject 3: "))
sub4 = float(input("Enter marks for Subject 4: "))
sub5 = float(input("Enter marks for Subject 5: "))

# Calculate total and percentage
total = sub1 + sub2 + sub3 + sub4 + sub5
percentage = (total / 500) * 100

print("Total Marks:", total)
print("Percentage:", percentage)

# Pass/Fail using logical operator
if (percentage >= 40) and (sub1 >= 35) and (sub2 >= 35) and (sub3 >= 35) and (sub4 >= 35) and (sub5 >= 35):
    print("Result: PASS")
else:
    print("Result: FAIL")
```

```python
## Check if a given email exists in a list of registered users.

registered_users = [
    "amit@gmail.com",
    "priya@yahoo.com",
    "rahul@hotmail.com",
    "neha@gmail.com",
    "rohit@outlook.com"
]

email = input("Enter your email address: ")

if email in registered_users:
    print("Login Successful! Email is registered.")
else:
    print("Email not found. Please register first.")
```

```python
## Build a simple calculator using arithmetic operators.

num1 = int(input("Enter 1st Number :"))
num2 = int(input("Enter 2nd Number :"))

operation = input("Enter operation to be perform +,-,*,/ :")

if (operation == "+"):
    print("Sum of number is", num1 + num2)
elif (operation == "-"):
    print("Sub of number is", num1 - num2)
elif (operation == "*"):
    print("Multi of number is", num1 * num2)
else:
    print("Div of number is", num1 / num2)
```

```python
## Count how many numbers in a list are divisible by 5.

numbers = [10, 23, 45, 50, 67, 75, 82, 90, 100]

count = 0

# Check each number
for num in numbers:
    if num % 5 == 0:
        count += 1

print("Numbers:", numbers)
print("Count of numbers divisible by 5:", count)
```

```python
## Write a number guessing game using comparison operators.

# Secret number
secret_number = 25

# Take user input
guess = int(input("Guess the number (1-50): "))

# Compare using comparison operators
if guess == secret_number:
    print("🎉 Correct! You guessed the number.")
elif guess > secret_number:
    print("📈 Too High!")
else:
    print("📉 Too Low!")
```
